```python
# 1.
number = int(input("Enter a number:"))
if number % 3 == 0 and number % 5 == 0:
    print("FizzBuzz")
elif number % 3 == 0:
    print("Fizz")
elif number % 5 == 0:
    print("Buzz")
else:
    print(number)
```

```python
# 2.
total = int(input("Enter seconds: "))

hours = total // 3600
remaining = total % 3600
minutes = remaining // 60
seconds = remaining % 60

print(hours, "hours", minutes, "minutes", seconds, "seconds")
```

```python
# 3.
weight = float(input("Enter weight: "))
height = float(input("Enter height: "))

bmi = weight / (height ** 2)

print("BMI:", bmi)

if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Normal")
elif bmi < 30:
    print("Overweight")
else:
    print("Obese")
```

```python
# 4.
hour = int(input("Enter hour: "))

if hour < 0 or hour > 23:
    print("Invalid time")
elif hour >= 5 and hour <= 11:
    print("Good Morning")
elif hour >= 12 and hour <= 16:
    print("Good Afternoon")
elif hour >= 17 and hour <= 20:
    print("Good Evening")
else:
    print("Good Night")
```

```python
# 5.
n = int(input("Enter a three-digit number: "))

last = n % 10
first = n // 100

if first == last:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# 6.
cp = float(input("Enter cost price: "))
sp = float(input("Enter selling price: "))

if sp > cp:
    amount = sp - cp
    percentage = (amount / cp) * 100
    print("Profit:", amount)
    print("Profit percentage:", percentage)

elif cp > sp:
    amount = cp - sp
    percentage = (amount / cp) * 100
    print("Loss:", amount)
    print("Loss percentage:", percentage)

else:
    print("No profit no loss")
```

```python
# 7.
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

d = b ** 2 - 4 * a * c

if d > 0:
    print("Real and different")
elif d == 0:
    print("Real and equal")
else:
    print("Imaginary")
```

```python
# 8.
month = int(input("Enter month number: "))

if month == 2:
    year = int(input("Enter year: "))

    if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
        print("29 days")
    else:
        print("28 days")

elif month in [4, 6, 9, 11]:
    print("30 days")

elif month in [1, 3, 5, 7, 8, 10, 12]:
    print("31 days")

else:
    print("Invalid month")
```

```python
# 9.
age = int(input("Enter age: "))
height = float(input("Enter height: "))

if age < 10:
    print("Too young")
elif age > 60:
    print("Too old")
elif height < 140:
    print("Too short")
else:
    print("Allowed")
```

```python
# 10.
distance = float(input("Enter distance: "))
hour = int(input("Enter hour: "))

if distance <= 2:
    fare = 60
else:
    fare = 60 + (distance - 2) * 15

if hour >= 22 or hour < 5:
    fare = fare + (fare * 25 / 100)

print("Final fare:", fare)
```

```python
# 11.
x = int(input("Enter x: "))
y = int(input("Enter y: "))

if x == 0 and y == 0:
    print("Origin")
elif y == 0:
    print("X-axis")
elif x == 0:
    print("Y-axis")
elif x > 0 and y > 0:
    print("Quadrant 1")
elif x < 0 and y > 0:
    print("Quadrant 2")
elif x < 0 and y < 0:
    print("Quadrant 3")
else:
    print("Quadrant 4")
```

```python
# 12.
a = int(input("Enter a: "))
b = int(input("Enter b: "))

a = a + b
b = a - b
a = a - b

print("After swapping:", a, b)

if a > b:
    print("a is greater")
elif b > a:
    print("b is greater")
else:
    print("Both are equal")
```
