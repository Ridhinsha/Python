### **part A: while + break (stop when you find what you need)**


```python
# The daily sales of a shop for 5 days are given. Find the first day on which the sales were more than 2000.
#Print the day number and the sales, then stop checking.
sales = [1200, 1500, 900, 2500, 1800]
n =len(sales)
print(n)
i = 1

while i<=n:

    if sales[i]>2000:
        print("Highest sales day-day",i+1)
        print("Highest sales=",sales[i])
        break
    i+=1
```

```python
# The marks of 5 students are given. Find the first student who failed (marks below 35). Print the student number and the marks, then stop.
marks = [78, 65, 90, 28, 55]
n=len(marks)
print(n)
i=1

while i<=n:
    if marks[i]<35:
        print("failed student",i+1)
        print("Low mark:",marks[i])
        break
    i+=1
```

```python
 # Add the daily sales one by one. Stop as soon as the running total becomes 5000 or more. Print the day number and the total at that point.
sales = [1200, 1500, 900, 2000, 1800]
i = 0
total = 0
n= len(sales)
print(n)

while i<=n:
    total += sales[i]
    if total>=5000:
      print("day",i+1)
      print("total:",total)
      break
    i+=1
```

### **Part B: while + continue (skip bad or unwanted values)**


```python
# A temperature data set has missing values stored as None. Print only the valid temperatures and skip the missing ones.
temperatures = [31, None, 34, None, 29]
i = 0
while i<len(temperatures):
    if temperatures[i]==None:
        i+=1
        continue
    print(temperatures[i])
    i+=1
```

```python
# Calculate the total of only the positive values in the list. Skip the negative values.
profit = [500, -200, 800, -100, 300]
i=0
total=0
while i<len(profit):
    total += profit[i]
    if profit[i]<0:
        i+=1
        continue
    print("postive values:",total)
    i+=1 
```

```python
# .The list shows the number of orders placed by 5 customers. Count how many customers placed at least one order. Skip the customers with 0 orders.
orders = [3, 0, 5, 0, 2]

i = 0
count = 0

while i < len(orders):
    if orders[i] == 0:
        i += 1
        continue
    count += 1
    i += 1

print(count)
```

### **Part C: while + break + continue together**

```python
# A data list has missing values (None), and the value -1 marks the end of the data. Skip the None values, stop when -1 is found, and print the total of the valid values before it.
data = [100, None, 250, None, 300, -1, 400]

i = 0
total = 0

while i < len(data):
    if data[i] == -1:
        break
    if data[i] is None:
        i += 1
        continue
    total = total + data[i]
    i += 1

print(total)
```

```python
# The shop was closed on some days, and the sales for those days are stored as 0. Skip the days with 0 sales and print the other sales. Stop the loop when a sale of more than 2000 is found, and print "Target crossed".
sales = [1200, 0, 1500, 0, 2500, 1800]

i = 0

while i < len(sales):
    if sales[i] == 0:
        i += 1
        continue
    if sales[i] > 2000:
        print("Target crossed")
        break
    print(sales[i])
    i += 1
```

```python
# .Write a program that keeps asking the user to enter a sale amount and adds it to the total:
#•  If the user enters a negative number, print "Invalid amount" and skip it. (Use continue.)
#•  If the user enters 0, stop the program. (Use break.)
#•  Otherwise, add the amount to the total.
# At the end, print the total sales. (Hint: use while True:)
total = 0

while True:
    amount = float(input("Enter sale amount: "))

    if amount < 0:
        print("Invalid amount")
        continue

    if amount == 0:
        break

    total = total + amount

print("Total sales:", total)
```

### **Part D: while with else + break**

```python
# .A list of customer IDs is given. Take an ID from the user and search for it using a while loop.
ids = [101, 102, 103, 104]

search_id = int(input("Enter customer ID: "))

i = 0

while i < len(ids):
    if ids[i] == search_id:
        print("Position:", i + 1)
        break
    i += 1
else:
    print("Customer not found")
```
