## print() 
- function is a built-in tool used to display text, variables, or other data outputs onto the screen

### Variable : 
            variable is a named pointer or "container" used to store data values in a computer's memory.
        
Variable creation :

```python
name = "Ridhinsha"
print(name)
Mark = 45
print(Mark)
```

Data types :
1. integer
   - used to store whole numbers
   - it does not contain decimal values 

```python
age = 25 
print(age)
```

2. float
    - used to store decimal numbers 

```python
Salary = 25000.00
print(Salary)
```

3. String
   - used to store text or characters
   - string are written inside quotes 

```python
name = "Amruth"
print(name)
```

4. Boolean
   - used to store true or false values

```python
is_student = True 
print(is_student)
```

5. list
  -  Used to store multiple values in a single variable.
  -  Values can be changed.

```python
Fruits = ["apple","Banana","Orange"]
print("data type of fruits:",type(Fruits))
```

6. tuple
   - used to  store multiple values
   - values cannot be changed after creation

```python
Mark = (12,13,14)
print(Mark)
```

7. set
   - used to store unique values
   - duplicates values  are automatically removed

```python
numbers = {1,2,3,4}
print(numbers)
```

8. dictionary
   - used to store data in key-values pairs

```python
student = {"Name":"Ridhinsha","Age":18,"course":"DA"}
print(student)
```
