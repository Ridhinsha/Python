### **Warm-up**

```python
book = {"title":"Python Basics","price": 300}
book.update({"autor":"Ravi"})
print(book)
```

```python
book = {"title":"Python Basics", "price": 300}
book.update({"pages":180})
book["price"] = 250
print(book)
```

```python
book = {"title":"python Basics","price":250,"pages":180}
book.pop("pages")
print(book)
```

```python
fruits = {"apple": 50, "banana": 30,"mango": 20}
removed = fruits.popitem()
print("Removed pair:", removed)
print("Remaining dictionary:", fruits)
```

```python
fruits = {"apple": 50, "banana": 30, "mango": 20}
copy_fruits = fruits.copy()
fruits.clear()
print("Original:", fruits)
print("Copy:", copy_fruits)
```

```python
d = {"a": 1, "b": 2, "c": 3}
d.pop("b")
d.popitem()
print(d)
```

### **Apply it**

```python
stock = {"apple": 50, "banana": 30, "mango": 20}
stock.update({"apple": 45, "orange": 25})
print(stock)
```

```python
fruits = {"apple": 50, "banana": 30, "mango": 20}
key = input("Enter a key to delete: ")
removed = fruits.pop(key, "Key not found")
print(removed)
```

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 20, "c": 3}
dict1.update(dict2)
print(dict1)
```

```python
a = {"x": 1}
b = a
b["x"] = 99
print(a)
```

```python
a = {"x": 1}
b = a.copy()
b["x"] = 99
print(a)
print(b)
```

```python
marks = {"Maths": 80, "Science": 75, "English": 90}
while marks:
    item = marks.popitem()
    print(item)
```

```python
cart = {"apple": 2, "banana": 3, "mango": 1}
backup = cart.copy()
cart.clear()
print("Cart:", cart)
print("Backup:", backup)
print("Length of cart:", len(cart))
print("Length of backup:", len(backup))
```

### **challenge**

```python
record = {}
while True:
    key = input("Enter key: ")

    if key == "stop":
        break

    value = input("Enter value: ")
    record.update({key: value})
```

```python
sentence = "to be or not to be"
words = sentence.split()
count = {}
for word in words:
    count.update({word: count.get(word, 0) + 1})
print(count)
```
