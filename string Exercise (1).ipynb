### **Mini Task - students ID card generator**

```python
data = " rahul kumar,data analytics,rahul.kumar@gmail.com,9876543210 "
# Remove extra spaces
data = data.strip()
# Split the data
details = data.split(",")
# Get each field
name = details[0].strip().title()
course = details[1].strip().title()
email = details[2].strip()
phone = details[3].strip()
# Username
at_position = email.find("@")
username = email[:at_position]
# Mask phone number
masked_phone = phone[:2] + "*" * 6 + phone[-2:]
# Initials
name_parts = name.split()
initials = ".".join(part[0].upper() for part in name_parts)
# Course code
course_code = course[:3].upper() + "-2026"
# Print Student ID Card
print("========== STUDENT ID CARD ==========")
print("Name      :", name)
print("Course    :", course)
print("Email     :", email)
print("Username  :", username)
print("Phone     :", masked_phone)
print("Initials   :", initials)
print("Course Code:", course_code)
print("=====================================")
```
