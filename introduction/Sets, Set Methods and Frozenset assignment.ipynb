### **Question 2: remove() vs discard()** ###

```python
# a- "green" remove
colors = {"Red", "Green", "Blue"}
colors.remove("Green") # Green remove

print(colors)
```

```python
# b- "yellow" remove
colors = {"Red", "Green", "Blue"}
colors.remove("Yellow") 
print(colors)
```

```python
# c- "yellow" discard
colors = {"Red", "Green", "Blue"}
colors.discard("Yellow")
print(colors)
```

```python
# d- own words 
I use remove() when I am sure that the item exists in the set.
If the item does not exist, remove() gives an error.
I use discard() when I am not sure whether the item exists, because discard() does not give an error if the item is missing.
```

### **Question 3: Set Operations** ###

```python
python_students = {"Arun", "Meera", "Rahul", "Sneha", "Kiran"}
sql_students = {"Meera", "Kiran", "Anil", "Divya"}

print(python_students.intersection(sql_students)) 
print(python_students.union(sql_students))
print(python_students.difference(sql_students))
print(python_students.symmetric_difference(sql_students))
```

### **Question 5: Set vs Frozenset (Concept + Code)** ###

```python
# a-
s = {1, 2, 3}
x = {s,4}
```

```python
# b-
s = frozenset({1, 2, 3})
x = {s,4}
print(x)
```

```python
# c-
A frozenset can be used as a dictionary key or as an element inside another set because it is immutable and hashable. A normal 
set cannot be used because it is mutable and unhashable.
```

```python
# d- real life use case
set: a set can be used to store unique student IDs ad remove duplicate IDs.
frozenset: A frozenset can be used to store fixed categories or permissions that should not be changed.
```
