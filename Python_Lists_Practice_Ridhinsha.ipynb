```python
## D1. Shopping List
shopping = ["milk", "bread", "eggs"]
shopping.append("butter")
shopping.insert(1, "rice")
shopping.extend(["salt", "sugar"])
shopping.remove("bread")
print(shopping)
print(shopping.index("eggs"))
shopping.sort()
print(shopping)
```

```python
## D2. Student Marks
marks = [72, 85, 90, 65, 88, 76, 95, 58]
marks.sort(reverse=True)
print(marks)
print("Top 3:", marks[:3])
print("Bottom 3:", marks[-3:])
total = sum(marks)
average = total / len(marks)
print("Total:", total)
print("Average:", average)
```

```python
## D3.Monthly sales report
months = ["jan","feb","mar","apr","may","jun"]
sales = [12000,15000,9000,18000,21000,17000]
print("Total:",sum(sales)) #sum()-Total
average = sum(sales)/len(sales) # len-function to count the length
print("average:",round(average,2)) # round-To round the value in fixed decimal place.
high_sale = max(sales) # max - high sales
print(high_sale)
position = sales.index(high_sale) # sales positions 
print("Best Month:",months[sales.index(high_sale)],"(",high_sale,")") # Best month
```

```python
count = 0
for sale in sales:
    if sale>= average:
        count +=1
print("months above average:",count)
```

```python
## D4. Search in a list
num =[4,7,2,7,9,7,1]
user_input = int(input("enter your number:"))
if user_input in num:
    print(user_input,"is",num.count(user_input),"times repeated")
    print("postion is:",num.index(user_input)+1)
else:
    print(user_input,"is not belongs to num")
```

```python
##D5. Remove Duplicates
data = [3, 1, 3, 2, 1, 5, 2]
new_list = []
for item in data:
    if item not in new_list:
        new_list.append(item)
print(new_list)
```

```python
## D6. Even and Odd Separator
numbers = [12, 7, 8, 3, 20, 15, 6, 9]
even = []
odd = []
for n in numbers:
    if n % 2 == 0:
        even.append(n)
    else:
        odd.append(n)
even.sort()
odd.sort()
print("Even:", even)
print("Odd:", odd)
```

```python
## D7. Palindrome List
a = [1, 2, 3, 2, 1]
if a == a[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
b = ["a", "b", "c"]
if b == b[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
## D8.Rotate a List
nums = [1, 2, 3, 4, 5, 6]
k = int(input("Enter k: "))
left = nums[k:] + nums[:k]
right = nums[-k:] + nums[:-k]
print("Left :", left)
print("Right:", right)
```

### **To-Do List**

```python
tasks = []

while True:
    print("\n1. Add task")
    print("2. Remove task")
    print("3. View tasks")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        task = input("Enter task: ")
        tasks.append(task)

    elif choice == "2":
        task = input("Enter task to remove: ")

        if task in tasks:
            tasks.remove(task)
        else:
            print("Task not found")

    elif choice == "3":
        print(tasks)

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice")
```
